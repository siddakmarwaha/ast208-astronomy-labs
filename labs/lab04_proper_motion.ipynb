Now that you have identified which source is the asteroid, you can measure its proper motion. <i class="fa fa-cogs" style="font-size:1.5em; color:red"></i> 6) What is the angular separation of the asteroid's positions between the two images? Calculate this in the code cell below, using Python as a calculator. Give the angular separation in units of arcseconds. Clearly print your answer with units.

<span style="color:red">(5 points total)</span>

```python
import numpy as np
#An exapmle of calculating a square root in python:
print(np.sqrt(4))

# An example of calculating a cosine in python
xx = np.cos(3.14159)
print(xx)
```

```python
d1 = 28+32/60+3.624/3600
ra1 = (3+58/60+00.5697/3600)*15
d2 = 28+32/60+01.987/3600
ra2 = (3+58/60+02.2353/3600)*15
```

```python
d1 = np.radians(d1)
ra1 = np.radians(ra1)
d2 = np.radians(d2)
ra2 = np.radians(ra2)
```

```python
costheta = np.cos(d1)*np.cos(d2)*(np.cos(ra1)*np.cos(ra2)+np.sin(ra1)*np.sin(ra2))+np.sin(d1)*np.sin(d2)
```

```python
theta = np.arccos(costheta)
theta
```

```python
print("The Angular Separation is: ", theta, 'radians')
```

```python
distance = 2.7 #AU
distance_km = distance * 1.496e+8
time_s = (time*3600)
pm_ast_kms = theta/time_s
transverse_vel = distance_km * np.tan(pm_ast_kms)
print("The Transverse Velocity is: ", transverse_vel, 'km/s')
```
